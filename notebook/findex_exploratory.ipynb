```python
# data information
df.info() #verbose=True
```

```python
# missing data
total = df.isnull().sum().sort_values(ascending=False)
percent = (df.isnull().sum()/df.isnull().count()).sort_values(ascending=False)
missing_data = pd.concat([total, percent], axis=1, keys=['Total', 'Percent'])
missing_data.query('Total > 0')
```

```python
# variable female: 1 is female; 2 is not
df.query('age.isnull()')
```

```python
# list of total unique values per column
col_list = []
nunique_list = []

for col in df:
  col_list.append(col)
  nunique_list.append(df[col].nunique())

# list(zip(col_list, nunique_list))
df_nunique = pd.DataFrame(
    {'column': col_list,
     'nunique': nunique_list
     }
)

df_nunique.head()
```

## Data preparation

```python
# copy df
df_ = df.copy()

# the NaN for 'age' will be filled in with the average
df_['age'].fillna(df['age'].mean(), inplace = True)

# verify NaN values
df_['age'].isna().values.any()
```

```python
# select columns with nan percentage > 0.6

missing_drop = missing_data.query('Percent > 0.6').index[0:].to_list()
missing_drop
```

```python
# select columns with unique value == 1
nunique_drop = df_nunique.query('nunique == 1').iloc[:,0].to_list()
nunique_drop
```

```python
# non relevante variables to clustering
non_relevant = ['wpid_random']
```

```python
# concatenate lists
list_drop = list(set(missing_drop) | set(nunique_drop) | set(non_relevant))

# drop columns
df_ = df_.drop(list_drop, axis=1)
```

```python
print('The new dataframe has {} rows and {} columns.'.format(df_.shape[0], df_.shape[1]))
```

```python
# list unique values per column
value_list = []
unique_list = []

for col in df_:
  value_list.append(col)
  unique_list.append(df[col].unique())

# list(zip(col_list, unique_list))
df_unique = pd.DataFrame(
    {'column': value_list,
     'nunique': unique_list
     }
)

df_unique.sample(10)
```

```python
# fills in the NaN values with a default value of -1 because 0 is used as the answer to some questions
# preserve the categorical nature of the variables while providing a distinct label for missing values
df_.fillna(-1, inplace=True)

# converts the column to integers
df_int = df_.astype(int)

# https://pandas.pydata.org/docs/user_guide/categorical.html
# converts integers to categories
# df_cat = df_cat.astype('category')

df_int.sample(10)
```

```python
# is NaN values?
df_int.isna().values.any()
```

[**Income quintile group**](https://ec.europa.eu/eurostat/statistics-explained/index.php?title=Glossary:Income_quintile_group): In social statistics, income quintile groups are computed on the basis of the total equivalised disposable income attributed to each member of the household.

The data (of each person) are ordered according to the value of the total equivalised disposable income. Four cut-point values (the so-called quintile cut-off points) of income are identified, dividing the survey population into five groups equally represented by 20 % of individuals each:

The first quintile group represents 20 % of the population with the lowest income (an income smaller or equal to the first cut-off value), and the fifth quintile group represents the 20 % of population with the highest income (an income greater than the fourth cut-off value).
